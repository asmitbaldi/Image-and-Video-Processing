```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("../images/dog_grayscale.png", cv2.IMREAD_GRAYSCALE)

plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
mean = 0
sigma = 25

noise = np.random.normal(mean, sigma, image.shape)

noisy_image = image.astype(np.float32) + noise

noisy_image = np.clip(noisy_image, 0, 255)

noisy_image = np.uint8(noisy_image)
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(noisy_image, cmap="gray")
plt.title("Gaussian Noisy Image")
plt.axis("off")

plt.show()
```

```python
sp_noisy_image = image.copy()
noise_prob = 0.05
random = np.random.rand(*image.shape)
sp_noisy_image[random < noise_prob / 2] = 255
sp_noisy_image[
    (random >= noise_prob / 2) & (random < noise_prob)
] = 0
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(sp_noisy_image, cmap="gray")
plt.title("Salt and Pepper Noise")
plt.axis("off")

plt.show()
```

```python
mean_original_3 = cv2.blur(image, (3, 3))
mean_gaussian_3 = cv2.blur(noisy_image, (3, 3))
mean_sp_3 = cv2.blur(sp_noisy_image, (3, 3))

mean_original_5 = cv2.blur(image, (5, 5))
mean_gaussian_5 = cv2.blur(noisy_image, (5, 5))
mean_sp_5 = cv2.blur(sp_noisy_image, (5, 5))
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(mean_original_3, cmap="gray")
plt.title("Original - Mean 3x3")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(mean_gaussian_3, cmap="gray")
plt.title("Gaussian Noise - Mean 3x3")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(mean_sp_3, cmap="gray")
plt.title("Salt & Pepper - Mean 3x3")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
median_gaussian_3 = cv2.medianBlur(noisy_image, 3)
median_sp_3 = cv2.medianBlur(sp_noisy_image, 3)

median_gaussian_5 = cv2.medianBlur(noisy_image, 5)
median_sp_5 = cv2.medianBlur(sp_noisy_image, 5)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(noisy_image, cmap="gray")
plt.title("Gaussian Noise")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(median_gaussian_3, cmap="gray")
plt.title("Gaussian - Median 3x3")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(sp_noisy_image, cmap="gray")
plt.title("Salt & Pepper Noise")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(median_sp_3, cmap="gray")
plt.title("Salt & Pepper - Median 3x3")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
min_gaussian_3 = cv2.erode(noisy_image, np.ones((3, 3), np.uint8))
min_sp_3 = cv2.erode(sp_noisy_image, np.ones((3, 3), np.uint8))

min_gaussian_5 = cv2.erode(noisy_image, np.ones((5, 5), np.uint8))
min_sp_5 = cv2.erode(sp_noisy_image, np.ones((5, 5), np.uint8))

max_gaussian_3 = cv2.dilate(noisy_image, np.ones((3, 3), np.uint8))
max_sp_3 = cv2.dilate(sp_noisy_image, np.ones((3, 3), np.uint8))

max_gaussian_5 = cv2.dilate(noisy_image, np.ones((5, 5), np.uint8))
max_sp_5 = cv2.dilate(sp_noisy_image, np.ones((5, 5), np.uint8))
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(noisy_image, cmap="gray")
plt.title("Gaussian Noisy Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(min_gaussian_3, cmap="gray")
plt.title("Min Filter 3x3")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(max_gaussian_3, cmap="gray")
plt.title("Max Filter 3x3")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(min_gaussian_5, cmap="gray")
plt.title("Min Filter 5x5")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(sp_noisy_image, cmap="gray")
plt.title("Salt & Pepper Noisy Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(min_sp_3, cmap="gray")
plt.title("Min Filter 3x3")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(max_sp_3, cmap="gray")
plt.title("Max Filter 3x3")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(min_sp_5, cmap="gray")
plt.title("Min Filter 5x5")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(sp_noisy_image, cmap="gray")
plt.title("Original Noisy Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(min_sp_5, cmap="gray")
plt.title("Min Filter 5x5")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(max_sp_5, cmap="gray")
plt.title("Max Filter 5x5")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
sobel_x = cv2.Sobel(
    image,
    cv2.CV_64F,
    1, 0,
    ksize=3
)

sobel_y = cv2.Sobel(
    image,
    cv2.CV_64F,
    0, 1,
    ksize=3
)
```

```python
sobel_magnitude = np.sqrt(
    sobel_x**2 + sobel_y**2
)

sobel_magnitude = np.uint8(
    np.clip(sobel_magnitude, 0, 255)
)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(np.abs(sobel_x), cmap="gray")
plt.title("Sobel X")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(np.abs(sobel_y), cmap="gray")
plt.title("Sobel Y")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sobel_magnitude, cmap="gray")
plt.title("Gradient Magnitude")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Apply Laplacian filter
laplacian = cv2.Laplacian(
    image,
    cv2.CV_64F
)

# Convert for display
laplacian_display = np.uint8(
    np.clip(np.abs(laplacian), 0, 255)
)
```

```python
# Enhance image using Laplacian
enhanced_image = image.astype(np.float64) - laplacian

# Keep values between 0 and 255
enhanced_image = np.clip(
    enhanced_image,
    0,
    255
)

# Convert to uint8
enhanced_image = np.uint8(enhanced_image)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(laplacian_display, cmap="gray")
plt.title("Laplacian Output")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(enhanced_image, cmap="gray")
plt.title("Laplacian Enhanced")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Low-pass filtered image using Gaussian blur
low_pass = cv2.GaussianBlur(
    image,
    (3, 3),
    0
)

plt.imshow(low_pass, cmap="gray")
plt.title("Low-Pass Filtered Image")
plt.axis("off")
plt.show()
```

```python
# High-frequency component
high_frequency = (
    image.astype(np.float32) -
    low_pass.astype(np.float32)
)

# Shift for display
high_frequency_display = np.abs(high_frequency)

high_frequency_display = np.uint8(
    np.clip(high_frequency_display, 0, 255)
)

plt.imshow(high_frequency_display, cmap="gray")
plt.title("High-Frequency Component")
plt.axis("off")
plt.show()
```

```python
# Amplification factor
A = 3

# High Boost filtering
high_boost = (
    A * image.astype(np.float32)
    - low_pass.astype(np.float32)
)

# Keep values between 0 and 255
high_boost = np.clip(
    high_boost,
    0,
    255
)

# Convert to uint8
high_boost = np.uint8(high_boost)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(low_pass, cmap="gray")
plt.title("Low-Pass Filtered")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(high_frequency_display, cmap="gray")
plt.title("High-Frequency Component")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(high_boost, cmap="gray")
plt.title("High Boost Filtered (A=3)")
plt.axis("off")

plt.tight_layout()
plt.show()
```
