```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
```

```python
img = Image.open("../images/dog_grayscale.png").convert("L")

img_array = np.array(img)

print("Image Dimensions:", img_array.shape)
print("Image Height:", img_array.shape[0])
print("Image Width:", img_array.shape[1])
print("Minimum Pixel Value:", np.min(img_array))
print("Maximum Pixel Value:", np.max(img_array))

plt.figure(figsize=(6,6))
plt.imshow(img_array, cmap="gray")
plt.title("Input Grayscale Image")
plt.axis("off")
plt.show()
```

```python
img = Image.open("../images/dog_grayscale.png").convert("L")
img_array = np.array(img)

negative = 255 - img_array

negative_img = Image.fromarray(negative)
negative_img.save("../images/negative_numpy.png")

print("Negative image saved as '../images/negative_numpy.png'")

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img_array, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(negative, cmap="gray")
plt.title("Digital Negative")
plt.axis("off")

plt.show()
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

img = Image.open("../images/imgg.jpeg").convert("L")
img_array = np.array(img)

# Find minimum and maximum pixel values
rmin = np.min(img_array)
rmax = np.max(img_array)

print("Minimum Pixel Value:", rmin)
print("Maximum Pixel Value:", rmax)

#Min-Max Contrast Stretching
contrast = ((img_array - rmin) / (rmax - rmin) * 255).astype(np.uint8)

contrast_img = Image.fromarray(contrast)
contrast_img.save("../images/contrast_numpy.png")

print("Contrast stretched image saved as '../images/contrast_numpy.png'")

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img_array, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(contrast, cmap="gray")
plt.title("Contrast Stretched Image")
plt.axis("off")

plt.show()
```

```python
img = Image.open("../images/dog_grayscale.png").convert("L")
img_array = np.array(img)

threshold = 127

binary = np.where(img_array >= threshold, 255, 0).astype(np.uint8)

binary_img = Image.fromarray(binary)
binary_img.save("../images/binary_numpy.png")

print("Threshold Value:", threshold)
print("Binary image saved as '../images/binary_numpy.png'")

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img_array, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

plt.show()
```
