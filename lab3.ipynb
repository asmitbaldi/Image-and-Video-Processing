```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("../images/dog_grayscale.png", cv2.IMREAD_GRAYSCALE)

plt.imshow(image, cmap="gray")
plt.title("Grayscale Dog Image")
plt.axis("off")
plt.show()

x = 100
y = 150

print("Coordinates:", (x, y))
print("Intensity:", image[x, y])
```

```python
x = 100
y = 150
rows,cols = image.shape
neighbors_4 = []
neighbors_4.append((x-1, y))
neighbors_4.append((x+1, y))
neighbors_4.append((x, y+1))
neighbors_4.append((x, y-1))
print("Selected pixel:", (x,y))
print("4 neighbors:",neighbors_4)
```

```python
output = cv2.cvtColor(image, cv2.COLOR_GRAY2RGB)

# Selected pixel = red
output[x, y] = [255, 0, 0]

# 4-neighbors = green
for nx, ny in neighbors_4:
    output[nx, ny] = [0, 255, 0]

plt.figure(figsize=(8, 8))
plt.imshow(output)
plt.xlim(y-20, y+20)
plt.ylim(x+20, x-20)
plt.grid()
plt.show()
```

```python
x = 100
y = 150

rows, cols = image.shape

neighbors_8 = []

for dx in [-1, 0, 1]:
    for dy in [-1, 0, 1]:

        if dx == 0 and dy == 0:
            continue

        nx = x + dx
        ny = y + dy

        #Just Checking the boundaries
        if 0 <= nx < rows and 0 <= ny < cols:
            neighbors_8.append((nx, ny))

print("Selected Pixel:", (x, y))
print("8-Connected Neighbors:")

for p in neighbors_8:
    print(p)
```

```python
output = cv2.cvtColor(image, cv2.COLOR_GRAY2RGB)

# Selected pixel = red
output[x, y] = [255, 0, 0]

# 8-neighbors = green
for nx, ny in neighbors_8:
    output[nx, ny] = [0, 255, 0]

plt.figure(figsize=(8, 8))
plt.imshow(output)

plt.xlim(y - 20, y + 20)
plt.ylim(x + 20, x - 20)

plt.grid()
plt.title("8-Connectivity")
plt.show()
```

```python
p = (100, 150)
q = (200, 250)

x1, y1 = p
x2, y2 = q

euclidean = np.sqrt((x1 - x2)**2 + (y1 - y2)**2)

city_block = abs(x1 - x2) + abs(y1 - y2)

chessboard = max(abs(x1 - x2),abs(y1 - y2))

print("pixel P:", p)
print("pixel Q:", q)

print("Euclidean:", euclidean)
print("City Block:", city_block)
print("Chessboard:", chessboard)
```

```python
pairs = [((10, 20), (30, 40)),((10, 10), (10, 30)),((20, 20), (40, 20)),((10, 10), (30, 40))]

for p, q in pairs:

    x1, y1 = p
    x2, y2 = q

    euclidean = np.sqrt((x1 - x2)**2 + (y1 - y2)**2)

    city_block = abs(x1 - x2) + abs(y1 - y2)

    chessboard = max(abs(x1 - x2),abs(y1 - y2))

    print("P =", p, " Q =", q)
    print("Euclidean :", euclidean)
    print("City Block:", city_block)
    print("Chessboard:", chessboard)
```
